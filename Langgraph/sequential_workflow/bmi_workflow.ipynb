```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
from IPython.display import Image, display
```

```python
class BMIState(TypedDict):
    weight_kg: float
    height_m: float
    bmi: float
    category: str
```

```python
def calculate_bmi(state: BMIState) -> BMIState:
    weight = state["weight_kg"]
    height = state["height_m"]
    bmi = weight / (height ** 2)
    state["bmi"] = round(bmi, 2)
    return state
```

```python
def categorize_bmi(state: BMIState) -> BMIState:
    bmi = state["bmi"]
    if bmi < 18.5:
        state["category"] = "Underweight"
    elif 18.5 <= bmi < 24.9:
        state["category"] = "Normal weight"
    elif 25 <= bmi < 29.9:
        state["category"] = "Overweight"
    else:
        state["category"] = "Obesity"
    return state
```

```python
graph = StateGraph(BMIState)

# nodes
graph.add_node("calculate_bmi", calculate_bmi)
graph.add_node("categorize_bmi", categorize_bmi)

# edges
graph.add_edge(START, "calculate_bmi")
graph.add_edge("calculate_bmi", "categorize_bmi")
graph.add_edge("categorize_bmi", END)

# compile graph
workflow = graph.compile()
```

```python
# execute the graph
initial_state = {"weight_kg": 70, "height_m": 1.67, "bmi": 0, "category": ""}

final_state = workflow.invoke(initial_state)

print(final_state)
```
